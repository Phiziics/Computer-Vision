```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
```

```python
log_path = Path("../logs/gesture_log.csv")

df = pd.read_csv(log_path)

df.head()
```

```python
df.info()
```

```python
df["timestamp"] = pd.to_datetime(df["timestamp"])

df.head()
```

```python
gesture_counts = df["gesture"].value_counts()

gesture_counts
```

```python
plt.figure(figsize=(8, 5))
gesture_counts.plot(kind="bar")
plt.title("Gesture Detection Counts")
plt.xlabel("Gesture")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df["minute"] = df["timestamp"].dt.floor("min")

gestures_over_time = (
    df.groupby(["minute", "gesture"])
    .size()
    .reset_index(name="count")
)

gestures_over_time.head()
```

```python
pivot_df = gestures_over_time.pivot(
    index="minute",
    columns="gesture",
    values="count"
).fillna(0)

pivot_df.head()
```

```python
plt.figure(figsize=(10, 5))

for gesture in pivot_df.columns:
    plt.plot(pivot_df.index, pivot_df[gesture], label=gesture)

plt.title("Gesture Detections Over Time")
plt.xlabel("Time")
plt.ylabel("Count")
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
total_gestures = len(df)
unique_gestures = df["gesture"].nunique()
most_common_gesture = df["gesture"].value_counts().idxmax()
least_common_gesture = df["gesture"].value_counts().idxmin()

print(f"Total gesture changes logged: {total_gestures}")
print(f"Unique gestures detected: {unique_gestures}")
print(f"Most common gesture: {most_common_gesture}")
print(f"Least common gesture: {least_common_gesture}")
```
